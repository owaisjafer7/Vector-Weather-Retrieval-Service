```python
import lakebase

print(
    lakebase.run_query(
        """
        SELECT
            constraint_name,
            constraint_type
        FROM information_schema.table_constraints
        WHERE table_name='weather_documents'
        """
    )
)
```

```python
import lakebase

rows = lakebase.run_query(
    """
    SELECT source_type, COUNT(*) 
    FROM weather_documents
    GROUP BY source_type
    """
)

rows
```

```python
import lakebase

rows = lakebase.run_query(
    "SELECT COUNT(*) AS count FROM weather_documents"
)

print(rows)
```

```python
import lakebase

rows = lakebase.run_query(
    """
    SELECT id, COUNT(*) AS count
    FROM weather_documents
    GROUP BY id
    HAVING COUNT(*) > 1
    """
)

print(rows)
```
